## 4. Deduplicate Reviews — Keep Latest Review per Order

Given the order is unique but can have multiple reviews and The CEO's question is about delivery promises. The customer's final review after receiving the package will direct reflection of whether the delivery experience satisfied them or not.

```python
reviews['review_creation_date'] = pd.to_datetime(reviews['review_creation_date'])

reviews_deduped = (
    reviews
    .sort_values('review_creation_date', ascending=False)
    .drop_duplicates(subset='order_id', keep='first')
    .reset_index(drop=True)
)

print(f"Reviews before dedup: {len(reviews)}")
print(f"Reviews after dedup:  {len(reviews_deduped)}")
print(f"Rows removed:         {len(reviews) - len(reviews_deduped)}")
print(f"\nColumns kept: {reviews_deduped.columns.tolist()}")
```

## 5. Building the Master Dataset

```python
olist_master_data = (
    orders
    .merge(customers, on='customer_id', how='left')
    .merge(reviews_deduped, on='order_id', how='left')
    .merge(order_items.groupby('order_id').agg(
        product_id=('product_id', 'first'),
        freight_value=('freight_value', 'mean')
    ).reset_index(), on='order_id', how='left')
    .merge(products[['product_id', 'product_category_name']], on='product_id', how='left')
    .merge(category_translation, on='product_category_name', how='left')
)

# Must equal original orders count
print(f"Orders original:  {len(orders)}")
print(f"Master table:     {len(olist_master_data)}")
print(f"Row count match:  {len(orders) == len(olist_master_data)}")
print(f"\nColumns: {olist_master_data.columns.tolist()}")
```

# Story 2: The "Real" Delay Calculator


```python
olist_master_data.head()
```

```python
# Check what order statuses exist
olist_master_data['order_status'].value_counts()
```

```python
## Keep only delivered orders
delivered = olist_master_data[
    olist_master_data['order_status'] == 'delivered'
].copy()

delivered.head()
```

```python
# Convert date columns to datetime
date_cols = [
    'order_delivered_customer_date',
    'order_estimated_delivery_date',
    'order_purchase_timestamp'
]

for col in date_cols:
    delivered[col] = pd.to_datetime(delivered[col])
```

```python
delivered.info()
```

```python
print(f"Total orders:     {len(olist_master_data)}")
print(f"Delivered orders: {len(delivered)}")
print(f"Excluded orders:  {len(olist_master_data) - len(delivered)}")
print(f"\nDate columns converted: {date_cols}")
```

96,478 delivered orders but only 96,470 have an actual delivery date — 8 rows are marked "delivered" but have no delivery date.

```python
# Check the 8 delivered orders with no delivery date
missing_delivery = delivered[
    delivered['order_delivered_customer_date'].isnull()
]

len(missing_delivery)
```

```python
missing_delivery
```

```python
missing_delivery[['order_id', 'order_status',
                         'order_delivered_customer_date',
                         'order_estimated_delivery_date']]
```

```python
# clean the delivery droping the 8
# Drop the 8 delivered orders with no actual delivery date
delivered_clean = delivered[
    delivered['order_delivered_customer_date'].notna()
].copy()
```

```python
delivered_clean.info()
```

```python
# Calculate days difference
# Positive = delivered early (good)
# Negative = delivered late (bad)
delivered_clean['days_difference'] = (
    delivered_clean['order_estimated_delivery_date'] -
    delivered_clean['order_delivered_customer_date']
).dt.days


delivered_clean.head()
```

```python
# Classify delivery status
def classify_delivery(days):
    if days >= 0:
        return 'On Time'
    elif days >= -5:
        return 'Late'
    else:
        return 'Super Late'

delivered_clean['delivery_status'] = delivered_clean['days_difference'].apply(classify_delivery)

delivered_clean.head()
```
