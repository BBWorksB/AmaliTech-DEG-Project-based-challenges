## 4. Deduplicate Reviews — Keep Latest Review per Order

Given the order is unique but can have multiple reviews and The CEO's question is about delivery promises. The customer's final review after receiving the package will direct reflection of whether the delivery experience satisfied them or not.

```python
reviews['review_creation_date'] = pd.to_datetime(reviews['review_creation_date'])

reviews_deduped = (
    reviews
    .sort_values('review_creation_date', ascending=False)
    .drop_duplicates(subset='order_id', keep='first')
    .reset_index(drop=True)
)

print(f"Reviews before dedup: {len(reviews)}")
print(f"Reviews after dedup:  {len(reviews_deduped)}")
print(f"Rows removed:         {len(reviews) - len(reviews_deduped)}")
print(f"\nColumns kept: {reviews_deduped.columns.tolist()}")
```

## 5. Building the Master Dataset

```python
olist_master_data = (
    orders
    .merge(customers, on='customer_id', how='left')
    .merge(reviews_deduped, on='order_id', how='left')
    .merge(order_items.groupby('order_id').agg(
        product_id=('product_id', 'first'),
        freight_value=('freight_value', 'mean')
    ).reset_index(), on='order_id', how='left')
    .merge(products[['product_id', 'product_category_name']], on='product_id', how='left')
    .merge(category_translation, on='product_category_name', how='left')
)

# Critical check — must equal original orders count
print(f"Orders original:  {len(orders)}")
print(f"Master table:     {len(olist_master_data)}")
print(f"Row count match:  {len(orders) == len(olist_master_data)}")
print(f"\nColumns: {olist_master_data.columns.tolist()}")
```
